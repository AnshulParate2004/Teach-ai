# SEO Keyword Suggestion Agent - Solution Notebook
**Industry:** Digital Marketing

This notebook contains a concrete implementation for the **SEO Keyword Suggestion Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that acts as an automated SEO assistant. The user will input a topic and a list of competitor keywords, and the agent must generate a list of 5 unique long-tail keywords not found in the competitor list.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
topic = "Best running shoes for flat feet."
competitor_keywords = ["flat feet running", "pronation shoes", "arch support runners"]
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an SEO Marketing AI. Generate a list of 5 long-tail keywords that are highly relevant to the topic but not explicitly listed in the competitor keywords. Output as a JSON list."""

user_prompt = f'''Topic: {topic}\nCompetitors: {competitor_keywords}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
