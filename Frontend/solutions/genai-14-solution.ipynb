# Property Matcher Agent - Solution Notebook
**Industry:** Real Estate

This notebook contains a concrete implementation for the **Property Matcher Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that acts as an automated real estate matchmaker. The user will input client preferences (budget, location, etc.), and the agent must evaluate a list of available properties and return the best matches.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
client_preferences = {"budget_max": 450000, "bedrooms": 3, "location": "Suburbs"}
available_properties = [
    {"id": 1, "price": 420000, "beds": 3, "location": "Suburbs"},
    {"id": 2, "price": 500000, "beds": 4, "location": "Suburbs"},
    {"id": 3, "price": 380000, "beds": 2, "location": "Downtown"}
]
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Real Estate Matchmaker AI. Find the properties that best match the client's preferences. Output a JSON list of recommended property IDs and a brief reason why."""

user_prompt = f'''Preferences: {client_preferences}\nAvailable: {available_properties}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
