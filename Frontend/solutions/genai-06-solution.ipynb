# Financial Fraud Detection AI - Solution Notebook
**Industry:** Finance

This notebook contains a concrete implementation for the **Financial Fraud Detection AI**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically reviews financial transactions. The agent must compare a new transaction against the user's typical historical behavior, flag any anomalies, and allocate a fraud risk score (Low/Medium/High).

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
transaction = {"amount": 4500, "location": "Nigeria", "time": "03:15 AM", "merchant": "Crypto Exchange"}
user_history = {"typical_location": "New York, USA", "avg_transaction": 45.50}
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Fraud Detection AI. Analyze the transaction against the user's history and flag any anomalies. Output JSON with 'risk_level' (Low, Medium, High) and 'reasons'."""

user_prompt = f'''History: {user_history}\nTransaction: {transaction}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
