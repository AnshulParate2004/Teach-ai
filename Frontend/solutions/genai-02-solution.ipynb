# Leave & Attendance Query Agent - Solution Notebook
**Industry:** Human Resources

This notebook contains a concrete implementation for the **Leave & Attendance Query Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically processes employee HR queries. The user will input their question (e.g., about sick days), and your agent must retrieve data from their mock HR record and generate a polite, accurate response.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
employee_query = "How many sick days do I have left?"
employee_record = {"employee_id": 1024, "sick_days_balance": 4, "vacation_balance": 12}
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an HR Leave & Attendance Assistant. Answer the employee's query accurately using their provided HR record. Be polite and concise."""

user_prompt = f'''Query: {employee_query}\nRecord: {employee_record}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
