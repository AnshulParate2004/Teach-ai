# HR Skill Gap Detection Agent - Solution Notebook
**Industry:** Human Resources

This notebook contains a concrete implementation for the **HR Skill Gap Detection Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically processes a candidate's resume against a provided job description. The user will input the resume, and your agent must evaluate it, detect missing skills, and allocate a score or mark based on the candidate's alignment with the job requirements.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
job_description = "We need a Senior Backend Engineer with 5+ years of Python and AWS experience."
applicant_resume = "Software Engineer with 3 years of React, JavaScript, and Node.js experience."
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an HR Tech Analyst AI. Compare the applicant's resume against the job description. Output a JSON array containing the missing skills."""

user_prompt = f'''Job Description: {job_description}\n\nApplicant Resume: {applicant_resume}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
