# Bank Loan Application Reviewer - Solution Notebook
**Industry:** Banking

This notebook contains a concrete implementation for the **Bank Loan Application Reviewer**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically processes a user's loan application. The agent must evaluate the user's financial data against the bank's strict lending policy and output an approval decision with an allocated risk score.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
applicant_data = {"income": 65000, "credit_score": 580, "loan_amount": 25000, "existing_debt": 40000}
bank_policy = "Minimum credit score 620. Debt-to-income ratio must be below 45%."
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Bank Loan Review AI. Analyze the applicant data against the bank policy. Output a JSON response with 'status' (Approve/Deny) and 'explanation'."""

user_prompt = f'''Policy: {bank_policy}\nApplicant: {applicant_data}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
