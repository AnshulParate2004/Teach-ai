# Feedback Sentiment Analyzer - Solution Notebook
**Industry:** Customer Success

This notebook contains a concrete implementation for the **Feedback Sentiment Analyzer**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that processes incoming customer feedback automatically. The user will provide a list of reviews, and the agent must evaluate them and allocate a sentiment score (Positive/Negative/Neutral) to each.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
customer_feedback = [
    "The new dashboard is incredibly fast and intuitive!",
    "Support took 3 days to reply to my urgent ticket. Unacceptable.",
    "It's okay, does what I need but nothing special."
]
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Customer Insights AI. Analyze the sentiment of each feedback item. Output a JSON list where each item has 'sentiment' (Positive, Negative, Neutral) and 'key_theme'."""

user_prompt = f'''Feedback: {customer_feedback}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
