# Interview Scheduling Assistant - Solution Notebook
**Industry:** Talent Acquisition

This notebook contains a concrete implementation for the **Interview Scheduling Assistant**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically drafts interview scheduling emails. The agent must cross-reference a candidate's availability with the hiring manager's calendar and draft an email offering suitable slots.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
candidate_email = "I am available on Thursday after 2 PM EST."
hiring_manager_calendar = {"Thursday": {"1:00 PM EST": "Busy", "2:30 PM EST": "Free", "4:00 PM EST": "Free"}}
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Recruiting AI. Draft a polite email to the candidate offering the best available time slots from the hiring manager's calendar that match the candidate's availability."""

user_prompt = f'''Manager Calendar: {hiring_manager_calendar}\nCandidate: {candidate_email}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
