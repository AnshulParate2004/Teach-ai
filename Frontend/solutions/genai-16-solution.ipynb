# Expense Report Summarizer - Solution Notebook
**Industry:** Accounting

This notebook contains a concrete implementation for the **Expense Report Summarizer**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically audits expense reports. The user will input receipt text, and the agent must extract the line items, categorize them, and flag any violations against the corporate expense policy.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
receipt_text = "Uber Ride: $45.20. Starbucks: $12.50. Delta Airlines: $450.00."
expense_policy = "Meals must be under $50. Flights over $300 require manager approval."
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an Accounting AI. Extract the expenses, categorize them, and flag any policy violations. Output a structured JSON report."""

user_prompt = f'''Policy: {expense_policy}\nReceipt: {receipt_text}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
