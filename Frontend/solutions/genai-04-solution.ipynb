# AI Proposal Writer Agent - Solution Notebook
**Industry:** Sales

This notebook contains a concrete implementation for the **AI Proposal Writer Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically generates sales proposals. The user will input client requirements and company service offerings, and the agent must draft a professional, formatted business proposal matching the two.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
client_requirements = "Needs a new e-commerce website with Stripe integration and a custom admin dashboard."
company_offerings = "Web Development ($100/hr), Stripe Integration ($2000 flat), UI/Design ($3000 flat)."
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an Expert Sales Proposal Writer. Generate a professional business proposal summarizing our offerings that directly address the client's needs. Include estimated costs."""

user_prompt = f'''Client Needs: {client_requirements}\nOur Offerings: {company_offerings}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
