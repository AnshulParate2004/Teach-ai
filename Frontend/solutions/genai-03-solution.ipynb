# Procurement Renewal Alert Agent - Solution Notebook
**Industry:** Procurement

This notebook contains a concrete implementation for the **Procurement Renewal Alert Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically monitors vendor contracts. The agent must process the contract's expiration date, compare it to the current date, and determine whether to trigger a renewal alert to the procurement team.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
contract_details = {"vendor": "Acme Corp", "service": "Cloud Hosting", "expiration_date": "2026-10-15", "auto_renew": True, "cost": "$50,000"}
current_date = "2026-09-15"
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Procurement Alert AI. Determine if a contract renewal alert should be sent based on the expiration date. Output a JSON decision with 'send_alert' (boolean) and 'reason' (string)."""

user_prompt = f'''Today: {current_date}\nContract: {contract_details}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
