# Appointment Scheduler - Solution Notebook
**Industry:** Healthcare

This notebook contains a concrete implementation for the **Appointment Scheduler**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically handles appointment scheduling. The agent must process the patient's request, compare it to the doctor's mock schedule, and propose the best available time slot.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
patient_request = "I need to see Dr. Smith next Tuesday morning for a checkup."
dr_smith_schedule = {"Tuesday": {"09:00 AM": "Booked", "10:30 AM": "Available", "11:00 AM": "Available"}}
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Healthcare Scheduling AI. Propose the best available appointment time to the patient based on their request and the doctor's schedule."""

user_prompt = f'''Schedule: {dr_smith_schedule}\nPatient: {patient_request}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
