# 1. LCEL Chain Building with LangChain
**Industry:** IT Services

Build an LCEL chain that takes a technical topic, generates 3 factual points about it, and translates those points into French.

```
import os
from dotenv import load_dotenv

# Load your API key
load_dotenv()
# os.environ['GOOGLE_API_KEY'] = 'your-api-key-here'
```

```
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser

llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash")

# Step 1: Generate facts
fact_prompt = ChatPromptTemplate.from_template("Generate 3 factual points about {topic}.")

# Step 2: Translate
translate_prompt = ChatPromptTemplate.from_template("Translate the following text into French:\n\n{text}")

# Build the LCEL Chain
chain = (
    fact_prompt 
    | llm 
    | StrOutputParser() 
    | {"text": lambda x: x} 
    | translate_prompt 
    | llm 
    | StrOutputParser()
)

# Run the chain
result = chain.invoke({"topic": "Kubernetes autoscaling"})
print(result)
```
