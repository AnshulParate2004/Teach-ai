# Insurance Claim Status Agent - Solution Notebook

This notebook contains a concrete implementation for the **Insurance Claim Status Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that handles incoming customer queries regarding insurance claims. The agent must look up the claim status in a mock database and generate an appropriate response (e.g., reminding them of missing documents).

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
claim_id = "CLM-88392"
customer_message = "Has my claim been approved yet?"
system_record = {"claim_id": "CLM-88392", "status": "Under Review", "missing_docs": ["Police Report"]}
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an Insurance Customer Support AI. Check the system record and answer the customer's message. If documents are missing, kindly remind them."""

user_prompt = f'''Customer: {customer_message}\nRecord: {system_record}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
