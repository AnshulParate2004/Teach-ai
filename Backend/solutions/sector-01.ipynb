## Task 2: Data & Setup
**Objective:** Set up Python, design a basic agent skeleton with input/output flow.

*Note: Since we are not using an external API key (like OpenAI), we will build a local `MockAgent` class that simulates LLM generation.*


```
import json
import datetime
import random

class MockProductivityAgent:
    def __init__(self, name="SkillzzaBot"):
        self.name = name
        self.memory = {}
        self.active_goals = []
        print(f"[{self.name}] Initialization complete. Ready to assist.")
        
    def _simulate_llm(self, prompt):
        # A simple keyword-based router to simulate LLM understanding
        prompt_lower = prompt.lower()
        if "schedule" in prompt_lower or "meeting" in prompt_lower:
            return "I have drafted a calendar invite for your meeting. Would you like me to send it?"
        elif "remember" in prompt_lower or "my name" in prompt_lower:
            return "I have updated my memory with your preferences."
        elif "break down" in prompt_lower or "step" in prompt_lower:
            return "Here is a step-by-step breakdown:\n1. Phase 1\n2. Phase 2\n3. Phase 3"
        else:
            return f"I understand you are asking about '{prompt[:20]}...'. I am ready to help."

    def chat(self, user_input):
        print(f"User: {user_input}")
        response = self._simulate_llm(user_input)
        print(f"Agent: {response}\n")
        return response

# Test the basic scaffold
agent = MockProductivityAgent()
agent.chat("Can you schedule a meeting for tomorrow at 2 PM?")
agent.chat("Hello, what can you do?")
```

## Task 3: Build & Execute (Memory and Goal-Setting)
**Objective:** Integrate memory functionality to retain user info and add goal-setting logic to track tasks across sessions.


```
class MemoryAgent(MockProductivityAgent):
    def __init__(self):
        super().__init__(name="MemoryBot")
        self.user_profile = {"name": "Alex", "preferences": []}
        
    def add_memory(self, fact):
        self.user_profile["preferences"].append(fact)
        print(f"[Memory Updated] Saved: '{fact}'")
        
    def add_goal(self, goal_name, steps):
        goal = {"goal": goal_name, "steps": steps, "completed": 0}
        self.active_goals.append(goal)
        print(f"[Goal Added] {goal_name} with {len(steps)} steps.")
        
    def progress_goal(self, goal_name):
        for g in self.active_goals:
            if g["goal"] == goal_name:
                if g["completed"] < len(g["steps"]):
                    completed_step = g["steps"][g["completed"]]
                    g["completed"] += 1
                    print(f"[Goal Progress] Completed: {completed_step}. ({g['completed']}/{len(g['steps'])})")
                    if g["completed"] == len(g["steps"]):
                        print(f"[Goal Achieved] {goal_name} is fully complete!")
                return
        print("[Error] Goal not found.")

# Test Cases
smart_agent = MemoryAgent()

# 1. Test Memory
smart_agent.add_memory("Prefers meetings in the afternoon.")
smart_agent.add_memory("Do not schedule meetings on Fridays.")

# 2. Test Goal Tracking
smart_agent.add_goal("Q3 Marketing Launch", ["Draft brief", "Review designs", "Launch campaign"])
smart_agent.progress_goal("Q3 Marketing Launch")
smart_agent.progress_goal("Q3 Marketing Launch")
smart_agent.progress_goal("Q3 Marketing Launch")
```

## Task 5: Audit & Responsible AI
**Objective:** Implement filters for biased responses, user consent, and disclaimers.


```
class ResponsibleAgent(MemoryAgent):
    def __init__(self):
        super().__init__()
        self.user_consent_granted = False
        self.blocked_keywords = ["confidential", "password", "salary", "biased_term"]
        
    def request_consent(self):
        print("\n[Privacy Notice] To provide personalized assistance, this agent stores your conversation history locally.")
        print("Do you consent to data storage? (Simulating 'Yes')")
        self.user_consent_granted = True
        print("[System] Consent granted.\n")
        
    def chat_safely(self, user_input):
        if not self.user_consent_granted:
            print("[System Block] Please grant user consent before interacting.")
            return
            
        # 1. Bias & Privacy Filter
        for word in self.blocked_keywords:
            if word in user_input.lower():
                print(f"User: {user_input}")
                print(f"Agent: [Redacted] I cannot process requests containing sensitive or restricted terms ('{word}').")
                return
                
        # 2. Ambiguity Disclaimer
        print(f"User: {user_input}")
        if "predict" in user_input.lower() or "financial" in user_input.lower():
            print("Agent: Disclaimer - I am an AI assistant. This information is generated probabilistically and should not constitute professional financial advice.")
            print("Agent: Here is the probabilistic forecast...")
        else:
            print("Agent: Processing request safely...")

# Testing Responsible AI Features
safe_agent = ResponsibleAgent()

# Fails without consent
safe_agent.chat_safely("What is on my calendar?")

# Grant consent
safe_agent.request_consent()

# Trigger Privacy Filter
safe_agent.chat_safely("Store my password: password123")

# Trigger Disclaimer
safe_agent.chat_safely("Predict the financial outcome of this project.")
```
