# Task_2_Data_Setup
## Eda Loan Data


```python
import pandas as pd
import numpy as np

# ─── Generate a synthetic loan applicant dataset ───────────────────────────
np.random.seed(42)
n = 500

data = {
    "applicant_id": range(1, n + 1),
    "age": np.random.randint(21, 60, n),
    "gender": np.random.choice(["Male", "Female"], n),
    "income_monthly": np.random.randint(5000, 80000, n),
    "employment_type": np.random.choice(["Salaried", "Self-Employed", "Business", "Unemployed"], n, p=[0.4, 0.3, 0.2, 0.1]),
    "loan_amount": np.random.randint(10000, 500000, n),
    "loan_term_months": np.random.choice([12, 24, 36, 48, 60], n),
    "num_dependents": np.random.randint(0, 6, n),
    "credit_history_score": np.random.randint(300, 900, n),
    "existing_loans": np.random.randint(0, 5, n),
    "education_level": np.random.choice(["None", "Primary", "Secondary", "Graduate", "Postgraduate"], n),
}

df = pd.DataFrame(data)

# Create a label: eligible = 1 if credit score >= 600 AND income >= 15000 AND not unemployed
df["eligible"] = (
    (df["credit_history_score"] >= 600) &
    (df["income_monthly"] >= 15000) &
    (df["employment_type"] != "Unemployed") &
    (df["loan_amount"] <= df["income_monthly"] * 12)
).astype(int)

# Save dataset
df.to_csv("loan_applicants.csv", index=False)
print("Dataset saved: loan_applicants.csv")

# ─── Exploratory Data Analysis ─────────────────────────────────────────────
print("\n=== Dataset Shape ===")
print(df.shape)

print("\n=== Data Types ===")
print(df.dtypes)

print("\n=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Basic Statistics ===")
print(df.describe())

print("\n=== Eligibility Distribution ===")
print(df["eligible"].value_counts())
print(f"Approval Rate: {df['eligible'].mean():.1%}")

print("\n=== Approval by Employment Type ===")
print(df.groupby("employment_type")["eligible"].mean().round(2))
```
