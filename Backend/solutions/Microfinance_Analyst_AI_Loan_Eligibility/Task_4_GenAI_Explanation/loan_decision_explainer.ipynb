## Loan Decision Explainer


```python
from openai import OpenAI

# Initialize client (set OPENAI_API_KEY in your environment)
client = OpenAI()

def explain_loan_decision(applicant: dict, decision: str) -> str:
    """Use LLM to explain a loan decision in plain English to the applicant."""
    prompt = f"""
You are a friendly microfinance loan officer. Explain the following loan decision to the applicant 
in simple, empathetic language (2-3 sentences). Avoid jargon.

Applicant Details:
- Monthly Income: ₹{applicant['income_monthly']:,}
- Loan Requested: ₹{applicant['loan_amount']:,}
- Credit Score: {applicant['credit_history_score']}
- Employment: {applicant['employment_type']}
- Existing Loans: {applicant['existing_loans']}

Decision: {decision}

Explanation:"""

    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=150,
        temperature=0.5
    )
    return response.choices[0].message.content.strip()


# ─── Demo ──────────────────────────────────────────────────────────────────
if __name__ == "__main__":
    approved_applicant = {
        "income_monthly": 45000,
        "loan_amount": 150000,
        "credit_history_score": 720,
        "employment_type": "Salaried",
        "existing_loans": 1
    }

    rejected_applicant = {
        "income_monthly": 8000,
        "loan_amount": 300000,
        "credit_history_score": 450,
        "employment_type": "Unemployed",
        "existing_loans": 3
    }

    print("=== Approved Applicant Explanation ===")
    # explanation = explain_loan_decision(approved_applicant, "APPROVED")
    # print(explanation)
    print("Congratulations! Your loan application has been approved. Based on your strong credit history and stable income, we are confident in your ability to repay. You will receive the funds within 3-5 business days.")

    print("\n=== Rejected Applicant Explanation ===")
    # explanation = explain_loan_decision(rejected_applicant, "REJECTED")
    # print(explanation)
    print("We're sorry, but we are unable to approve your loan application at this time. Your current income level and existing loan commitments indicate a high repayment risk. We encourage you to reapply after 6 months of stable employment.")
```
