# Task_2_Data_Setup
## Data Preprocessing


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

np.random.seed(42)
n = 1000
departments = ['Waste Management','Public Works','Utilities','Property & Land','Traffic']
urgencies = ['High','Medium','Low']
complaint_types = [
    'Pothole on main road causing accidents',
    'Garbage not collected for 2 weeks',
    'Water supply cut without notice',
    'Illegal construction next door',
    'Broken streetlight near school',
]

df = pd.DataFrame({
    'complaint_id': [f'CMP{i:05d}' for i in range(1, n+1)],
    'complaint_text': np.random.choice(complaint_types, n),
    'department': np.random.choice(departments, n),
    'urgency': np.random.choice(urgencies, n, p=[0.22, 0.45, 0.33]),
    'status': np.random.choice(['Resolved','Pending','Escalated'], n, p=[0.55, 0.35, 0.10]),
})

le_dept = LabelEncoder()
le_urg = LabelEncoder()
df['dept_encoded'] = le_dept.fit_transform(df['department'])
df['urgency_encoded'] = le_urg.fit_transform(df['urgency'])
df.to_csv('complaints_clean.csv', index=False)
print('Cleaned dataset:', df.shape)
print(df['department'].value_counts())
```
