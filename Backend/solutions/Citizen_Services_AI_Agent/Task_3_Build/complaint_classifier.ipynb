## Complaint Classifier


```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

np.random.seed(42)
n = 1000
complaint_dept_map = [
    ('Pothole on main road causing accidents', 'Public Works'),
    ('Garbage not collected for 2 weeks', 'Waste Management'),
    ('Water supply cut without notice', 'Utilities'),
    ('Illegal construction next door', 'Property & Land'),
    ('Broken streetlight near school', 'Public Works'),
]
texts = [t for t, _ in complaint_dept_map] * 200
labels = [d for _, d in complaint_dept_map] * 200

X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=42)
pipeline = Pipeline([('tfidf', TfidfVectorizer()), ('clf', RandomForestClassifier(random_state=42))])
pipeline.fit(X_train, y_train)
preds = pipeline.predict(X_test)
print(classification_report(y_test, preds))

test_complaints = ['There is a large pothole damaging vehicles', 'No water since yesterday morning']
for c in test_complaints:
    dept = pipeline.predict([c])[0]
    print(f'Complaint: {c[:50]}\nDept: {dept}\n')
```
