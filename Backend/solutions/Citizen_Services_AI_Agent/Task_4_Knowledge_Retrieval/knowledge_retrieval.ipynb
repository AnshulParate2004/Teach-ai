```python
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain.schema import Document

RESOLUTION_KB = [
    Document(page_content='Pothole complaints: Field team deployed within 72 hours. Citizen notified by SMS. Work order generated in PWD system.', metadata={'dept': 'Public Works'}),
    Document(page_content='Garbage collection: Missed pickup rescheduled within 48 hours. Crew supervisor alerted. Apology SMS sent to citizen.', metadata={'dept': 'Waste Management'}),
    Document(page_content='Water supply disruption: Emergency tanker dispatched if outage > 24 hours. Utility team notified. Estimated restoration time communicated.', metadata={'dept': 'Utilities'}),
    Document(page_content='Illegal construction: Site inspection scheduled within 5 days. Show cause notice issued to property owner if violation confirmed.', metadata={'dept': 'Property & Land'}),
]

def get_resolution(complaint_text: str, department: str) -> str:
    for doc in RESOLUTION_KB:
        if doc.metadata['dept'] == department:
            return doc.page_content
    return 'Standard resolution process initiated. Ticket escalated to department head.'

tests = [('There is a large pothole', 'Public Works'), ('No water for 2 days', 'Utilities')]
for complaint, dept in tests:
    print(f'Complaint: {complaint}\nResolution: {get_resolution(complaint, dept)}\n')
```
