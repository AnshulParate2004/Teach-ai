## Insights Generator


```python
from openai import OpenAI
client = OpenAI()

def generate_insight(question, data_result):
    prompt = f"""You are a business analyst.
The user asked: "{question}"
Query result: {data_result}
Give a 2-sentence business insight and recommended action."""
    # response = client.chat.completions.create(model="gpt-3.5-turbo",
    #     messages=[{"role":"user","content":prompt}])
    # return response.choices[0].message.content
    return (f"The data shows {data_result}. "
            "This indicates a strong market opportunity that should be prioritized in the upcoming quarter's strategy.")

samples = [
    ("Which product is most profitable?", "Laptop with $4.2M profit margin at 62%"),
    ("Which region is underperforming?", "East region is 35% below target revenue"),
]

for q, r in samples:
    print(f"Q: {q}")
    print(f"Insight: {generate_insight(q, r)}")
    print()
```
