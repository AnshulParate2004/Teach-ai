# Task_2_Data_Setup
## Data Setup


```python
import pandas as pd
import numpy as np
import sqlite3

np.random.seed(42)
n = 1000
df = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=n, freq="D").strftime("%Y-%m-%d")[:n],
    "product": np.random.choice(["Laptop","Phone","Tablet","Watch","Headphones"], n),
    "category": np.random.choice(["Electronics","Accessories"], n),
    "region": np.random.choice(["North","South","East","West"], n),
    "units_sold": np.random.randint(1, 100, n),
    "revenue": np.random.randint(5000, 500000, n),
    "profit": np.random.randint(500, 100000, n),
})
df.to_csv("sales_data.csv", index=False)
conn = sqlite3.connect("sales.db")
df.to_sql("sales", conn, if_exists="replace", index=False)
conn.close()
print("Dataset ready:", df.shape)
print(df.head())
```
