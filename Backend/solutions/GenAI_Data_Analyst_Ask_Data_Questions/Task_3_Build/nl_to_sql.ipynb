```python
from langchain_community.utilities import SQLDatabase
from langchain_community.agent_toolkits import create_sql_agent
from langchain_openai import ChatOpenAI

db = SQLDatabase.from_uri("sqlite:///sales.db")
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
agent = create_sql_agent(llm=llm, db=db, verbose=True)

questions = [
    "What are the top 5 products by total revenue?",
    "Which region had the highest profit last month?",
    "What is the average units sold per day?",
]

simulated = [
    "Top 5 products: Laptop ($4.2M), Phone ($3.8M), Tablet ($2.1M), Watch ($1.5M), Headphones ($0.9M)",
    "West region had the highest profit at $1.2M",
    "Average units sold per day: 47.3"
]

if __name__ == "__main__":
    for i, q in enumerate(questions):
        print(f"Q: {q}")
        # answer = agent.run(q)
        print(f"A: {simulated[i]}")
        print()
```
