```python
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain.chains import RetrievalQAWithSourcesChain

DISCLAIMER = '\n\n*Consult a qualified healthcare provider for medical advice.*'

def ask_clinical_question(question: str) -> str:
    simulated = {
        'What treats Type 2 Diabetes?': 'Metformin is first-line. (Source: ADA 2024)',
        'What is hypertension?': 'BP consistently above 130/80 mmHg. (Source: AHA 2023)',
    }
    return simulated.get(question, 'Not found in knowledge base.') + DISCLAIMER

if __name__ == '__main__':
    for q in ['What treats Type 2 Diabetes?', 'What is hypertension?']:
        print(f'Q: {q}')
        print(f'A: {ask_clinical_question(q)}\n')
```
