## Prompt Testing


```python
from openai import OpenAI

client = OpenAI()

tickets = [
    {"id": "T001", "description": "My credit card was charged twice.", "priority": "High"},
    {"id": "T002", "description": "Dashboard loads slowly.", "priority": "Medium"},
    {"id": "T003", "description": "I want to add a team member.", "priority": "Low"},
]

def classify_ticket(description):
    # Simulated output (uncomment with real key)
    if "charged" in description or "invoice" in description:
        return "Billing"
    elif "loads" in description or "crash" in description:
        return "Technical"
    elif "add" in description or "account" in description:
        return "Account"
    return "Other"

def summarize_ticket(description, priority):
    return f"{description[:60].rstrip('.')} - {priority} priority issue requiring prompt attention."

print("=== Prompt Testing Results ===")
for t in tickets:
    cat = classify_ticket(t["description"])
    summary = summarize_ticket(t["description"], t["priority"])
    print(f"[{t['id']}] Category: {cat}")
    print(f"       Summary: {summary}")
```
