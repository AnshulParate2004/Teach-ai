# Refund Processing Agent - Solution Notebook
**Industry:** E-Commerce

This notebook contains a concrete implementation for the **Refund Processing Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically processes e-commerce refund requests. The agent must evaluate the customer's complaint against the company return policy and order details, determining whether they are eligible for a refund.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
customer_request = "My shirt arrived torn. I want my money back."
order_details = {"order_id": 992, "item": "Shirt", "days_since_delivery": 12, "condition": "Damaged"}
return_policy = "Full refund offered within 30 days for damaged items. Must provide photo."
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are an E-Commerce Support AI. Determine if the customer is eligible for a refund. Output JSON with 'eligible' (boolean) and 'next_steps' (string)."""

user_prompt = f'''Policy: {return_policy}\nOrder: {order_details}\nCustomer: {customer_request}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
