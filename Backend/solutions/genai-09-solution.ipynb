# Sales Lead Qualifier Agent - Solution Notebook
**Industry:** Sales Operations

This notebook contains a concrete implementation for the **Sales Lead Qualifier Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that processes incoming sales leads. The agent must compare the lead's profile to our Ideal Customer Profile (ICP) and automatically allocate a lead score out of 100, marking them as Hot, Warm, or Cold.

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
lead_data = {"name": "Jane Doe", "company_size": 500, "budget": "$10k+", "timeline": "Immediate"}
ideal_customer_profile = {"min_company_size": 100, "min_budget": "$5k", "timeline": "0-3 months"}
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Sales Lead Qualifier AI. Score the lead from 1-100 based on the Ideal Customer Profile. Output JSON with 'score' (integer) and 'qualification' (Hot, Warm, Cold)."""

user_prompt = f'''ICP: {ideal_customer_profile}\nLead: {lead_data}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
