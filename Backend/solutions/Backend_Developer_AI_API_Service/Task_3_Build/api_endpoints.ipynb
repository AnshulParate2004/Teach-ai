## Api Endpoints


```python
from fastapi import FastAPI, Header, HTTPException
from pydantic import BaseModel
from openai import OpenAI

app = FastAPI(title='AI API Service')
client = OpenAI()
VALID_KEYS = {'sk-demo-key-123'}

def verify_key(x_api_key: str = Header(...)):
    if x_api_key not in VALID_KEYS: raise HTTPException(401, 'Invalid API key')

class TextRequest(BaseModel):
    text: str

@app.post('/v1/summarize')
def summarize(req: TextRequest, key: str = Header(..., alias='X-API-Key')):
    verify_key(key)
    # r = client.chat.completions.create(model='gpt-3.5-turbo',
    #     messages=[{'role':'user','content':f'Summarize: {req.text}'}])
    return {'summary': f'Summary of: {req.text[:100]}...'}

@app.post('/v1/classify')
def classify(req: TextRequest, key: str = Header(..., alias='X-API-Key')):
    verify_key(key)
    categories = ['Billing','Technical','Account','Other']
    import random; category = random.choice(categories)
    return {'category': category, 'confidence': 0.87}
```
