## Campaign Optimizer


```python
import numpy as np

campaigns = {
    'Google Ads': {'budget': 360000, 'roas': 3.2, 'conversions': 284},
    'Facebook Ads': {'budget': 510000, 'roas': 1.8, 'conversions': 147},
    'Email Marketing': {'budget': 24000, 'roas': 7.4, 'conversions': 335},
}

total_budget = sum(c['budget'] for c in campaigns.values())
print(f'Total Budget: Rs {total_budget:,}')

# Weight allocation by ROAS efficiency
total_roas = sum(c['roas'] for c in campaigns.values())
print('\nOptimized Budget Allocation (ROAS-weighted):')
for name, data in campaigns.items():
    new_budget = total_budget * (data['roas'] / total_roas)
    projected_conv = int(data['conversions'] * (new_budget / data['budget']))
    print(f'  {name}: Rs {new_budget:,.0f} (projected conversions: {projected_conv})')
```
