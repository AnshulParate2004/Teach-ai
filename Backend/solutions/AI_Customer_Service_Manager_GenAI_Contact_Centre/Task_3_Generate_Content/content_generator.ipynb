## Content Generator


```python
from openai import OpenAI
client = OpenAI()

SEGMENTS = {
    'High-Value Loyalists': 'Exclusive early access to new products and VIP loyalty rewards.',
    'At-Risk Churners': 'A special win-back offer: 25% off your next purchase, just for you.',
    'New Explorers': 'Discover our best sellers with a welcome discount of 15%.',
    'Bargain Hunters': 'Flash sale today only: Up to 40% off selected items.',
}

def generate_email_copy(segment: str, product: str) -> dict:
    hook = SEGMENTS.get(segment, 'Check out our latest offers!')
    return {
        'subject': f'{segment}: {hook[:40]}',
        'body': (f'Hi there,\n\n{hook}\n\nShop {product} now at techcorp.com\n\nBest,\nThe TechCorp Team'),
        'cta': 'Shop Now'
    }

for segment in SEGMENTS:
    copy = generate_email_copy(segment, 'AI Productivity Suite')
    print(f'Segment: {segment}')
    print(f'Subject: {copy["subject"]}')
    print(f'Body preview: {copy["body"][:100]}...\n')
```
