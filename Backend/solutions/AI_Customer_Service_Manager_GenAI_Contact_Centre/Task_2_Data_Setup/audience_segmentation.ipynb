# Task_2_Data_Setup
## Audience Segmentation


```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans

np.random.seed(42)
n = 500
df = pd.DataFrame({
    'age': np.random.randint(18, 65, n),
    'avg_order_value': np.random.randint(500, 15000, n),
    'purchase_frequency': np.random.randint(1, 20, n),
    'email_opens': np.random.randint(0, 50, n),
    'days_since_last_visit': np.random.randint(1, 180, n),
})

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['segment'] = kmeans.fit_predict(df)

segment_names = {0: 'High-Value Loyalists', 1: 'At-Risk Churners', 2: 'New Explorers', 3: 'Bargain Hunters'}
df['segment_name'] = df['segment'].map(segment_names)

print('Audience Segments:')
print(df.groupby('segment_name').size())
print('\nSegment Profiles:')
print(df.groupby('segment_name')[['avg_order_value','purchase_frequency']].mean().round(1))
```
