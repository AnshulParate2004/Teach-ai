## Chatbot Config


```python
from openai import OpenAI
client = OpenAI()

ARIA_SYSTEM_PROMPT = """You are Aria, the virtual receptionist for InnoSpace Coworking.
Your personality is warm, professional, and helpful.
You help visitors with: operating hours, membership plans, room bookings, directions, and FAQs.
Keep responses to 2-3 sentences. If unsure, offer to connect with a human staff member."""

def aria_respond(user_message: str, history: list = None) -> str:
    messages = [{'role': 'system', 'content': ARIA_SYSTEM_PROMPT}]
    if history: messages.extend(history)
    messages.append({'role': 'user', 'content': user_message})
    # r = client.chat.completions.create(model='gpt-3.5-turbo', messages=messages, max_tokens=100)
    # return r.choices[0].message.content
    responses = {
        'hours': 'We are open Monday to Saturday, 8 AM to 10 PM. Welcome anytime!',
        'price': 'Our plans start at Rs 499 for a Day Pass. Monthly Hot Desks are Rs 5,999.',
        'parking': 'Two-wheeler parking is free! Four-wheeler is Rs 50/hour.',
    }
    for k, v in responses.items():
        if k in user_message.lower(): return v
    return 'Welcome to InnoSpace! I can help with hours, plans, bookings, and directions.'

queries = ['What are your hours?', 'What does a desk cost?', 'Is parking free?']
for q in queries:
    print(f'Visitor: {q}\nAria: {aria_respond(q)}\n')
```
