## Sentiment Analysis


```python
import pandas as pd
import numpy as np
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

sample_tweets = [
    ('FitSphere', 'Love my new resistance bands! Great quality and fast delivery!'),
    ('FitSphere', 'The packaging was damaged when it arrived. Very disappointed.'),
    ('HomeFit', 'Decent product but overpriced compared to FitSphere.'),
    ('ActiveEdge', 'Best home gym equipment I have ever bought. Highly recommend!'),
    ('FitSphere', 'Average quality for the price. Nothing special.'),
]

results = []
for brand, text in sample_tweets:
    scores = analyzer.polarity_scores(text)
    sentiment = 'positive' if scores['compound'] > 0.05 else ('negative' if scores['compound'] < -0.05 else 'neutral')
    results.append({'brand': brand, 'text': text[:60], 'sentiment': sentiment, 'score': round(scores['compound'], 3)})

df = pd.DataFrame(results)
print(df.to_string(index=False))
print('\nFitSphere Sentiment:', df[df.brand=='FitSphere']['sentiment'].value_counts().to_dict())
```
