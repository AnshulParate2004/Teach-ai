## Data Collection


```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 500
brands = ['FitSphere', 'HomeFit', 'ActiveEdge']
sentiments = ['positive', 'negative', 'neutral']
topics = ['product quality', 'delivery', 'pricing', 'customer service', 'workout results']

df = pd.DataFrame({
    'tweet_id': range(1, n+1),
    'brand': np.random.choice(brands, n, p=[0.5, 0.3, 0.2]),
    'text': [f'Sample tweet about {np.random.choice(brands)} {np.random.choice(topics)}' for _ in range(n)],
    'likes': np.random.randint(0, 500, n),
    'retweets': np.random.randint(0, 100, n),
    'date': pd.date_range('2024-01-01', periods=n, freq='4H')[:n],
})

df.to_csv('social_media_data.csv', index=False)
print('Dataset created:', df.shape)
print(df['brand'].value_counts())
```
