## Theme Extractor


```python
from openai import OpenAI
client = OpenAI()

def extract_themes(tweets: list) -> list:
    combined = '\n'.join(f'- {t}' for t in tweets[:20])
    prompt = (f'Identify the top 5 recurring themes in these customer tweets.\n\n{combined}\n\nThemes:')
    # r = client.chat.completions.create(model='gpt-3.5-turbo', messages=[{'role':'user','content':prompt}])
    # return r.choices[0].message.content.split('\n')
    return [
        '1. Product Quality - customers praise durability and build quality',
        '2. Delivery Speed - fast shipping praised, damaged packaging criticized',
        '3. Value for Money - mixed views; some find it priced fairly, others expensive',
        '4. Workout Results - many users reporting visible fitness improvements',
        '5. Customer Service - quick response times highlighted positively',
    ]

sample_tweets = ['Great quality bands!', 'Delivery was super fast', 'A bit pricey but worth it']
themes = extract_themes(sample_tweets)
print('Top Themes:')
for t in themes: print(t)
```
