# 12. Building a Custom MCP Server (File Manager)
**Industry:** Telecommunications

Build a Model Context Protocol (MCP) server that exposes file operations (read, write) as tools an LLM client can call.

```
!pip install mcp
```

```
from mcp.server.fastmcp import FastMCP

# Create an MCP server
mcp = FastMCP("File Manager")

@mcp.tool()
def read_file(path: str) -> str:
    """Reads a network config file"""
    try:
        with open(path, 'r') as f:
            return f.read()
    except Exception as e:
        return str(e)

@mcp.tool()
def write_file(path: str, content: str) -> str:
    """Writes to a network config file"""
    try:
        with open(path, 'w') as f:
            f.write(content)
        return "Success"
    except Exception as e:
        return str(e)

print("Run this script directly via stdio transport: `uv run mcp_server.py`")
# mcp.run(transport='stdio')
```
