```python
from openai import OpenAI
client = OpenAI()

def llm_classify(query: str) -> str:
    prompt = f'Classify this support query into one of [billing, technical, faq, escalation]:\n"{query}"\nCategory:'
    # r = client.chat.completions.create(model='gpt-3.5-turbo',
    #     messages=[{'role':'user','content':prompt}], max_tokens=10)
    # return r.choices[0].message.content.strip().lower()
    if 'bill' in query.lower(): return 'billing'
    if 'crash' in query.lower(): return 'technical'
    return 'faq'

def llm_respond(intent: str, query: str) -> str:
    prompts = {
        'billing': f'You are a billing agent. Answer: {query}',
        'technical': f'You are a tech support agent. Answer: {query}',
        'faq': f'You are a helpful assistant. Answer: {query}',
    }
    return f'[LLM {intent.upper()} response to: {query[:50]}]'

queries = ['My invoice is incorrect','App crashed on startup','What is your cancellation policy?']
for q in queries:
    intent = llm_classify(q)
    response = llm_respond(intent, q)
    print(f'Q: {q}\nIntent: {intent}\nResponse: {response}\n')
```
