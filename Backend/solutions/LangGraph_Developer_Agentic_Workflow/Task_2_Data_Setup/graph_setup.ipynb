## Graph Setup


```python
from typing import TypedDict, Literal
from langgraph.graph import StateGraph, END

class SupportState(TypedDict):
    query: str
    intent: str
    response: str
    escalate: bool

def classify_node(state: SupportState) -> SupportState:
    q = state['query'].lower()
    if any(w in q for w in ['bill','charge','refund','payment']):
        state['intent'] = 'billing'
    elif any(w in q for w in ['crash','error','bug','slow']):
        state['intent'] = 'technical'
    elif any(w in q for w in ['human','agent','speak','manager']):
        state['intent'] = 'escalation'
    else:
        state['intent'] = 'faq'
    return state

def route_intent(state: SupportState) -> Literal['billing','technical','faq','escalation']:
    return state['intent']

print('LangGraph nodes and state defined.')
print('Nodes: classify_node, billing_node, technical_node, faq_node, escalation_node')
```
