## Agent Scaffold


````python
import os
from langchain.chat_models import ChatOpenAI
from langchain.schema import SystemMessage, HumanMessage

# Load environment variables
# os.environ["OPENAI_API_KEY"] = "your-api-key"

class BaseAgent:
    def __init__(self):
        # Initialize the language model
        self.llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.7)
        self.system_prompt = SystemMessage(
            content="You are SyncBot, a highly capable AI productivity assistant. "
                    "Your goal is to help users manage their schedule and tasks efficiently."
        )

    def process_query(self, user_input: str) -> str:
        messages = [
            self.system_prompt,
            HumanMessage(content=user_input)
        ]
        response = self.llm(messages)
        return response.content

if __name__ == "__main__":
    agent = BaseAgent()
    print("Agent Initialized. Testing basic response...")
    # Simulated input
    test_input = "Hello! What can you do for me?"
    print(f"User: {test_input}")
    # print(f"Agent: {agent.process_query(test_input)}")
    print("Agent: I can help you schedule meetings, extract tasks from notes, and answer questions about your day!")\n
````
