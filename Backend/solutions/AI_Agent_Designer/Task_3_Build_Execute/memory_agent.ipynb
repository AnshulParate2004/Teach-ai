## Memory Agent


````python
import os
from langchain.chat_models import ChatOpenAI
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain

class MemoryAgent:
    def __init__(self):
        # Setup LLM and memory
        self.llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
        self.memory = ConversationBufferMemory()
        self.chain = ConversationChain(
            llm=self.llm,
            memory=self.memory,
            verbose=False
        )
        self.goals = []

    def set_goal(self, goal: str):
        self.goals.append(goal)
        print(f"[Goal Added] {goal}")

    def chat(self, user_input: str) -> str:
        # Check goals
        context = ""
        if self.goals:
            context = f"(Active Goals: {', '.join(self.goals)}) "
            
        response = self.chain.predict(input=context + user_input)
        return response

if __name__ == "__main__":
    agent = MemoryAgent()
    print("--- Memory & Goal-Setting Test ---")
    
    agent.set_goal("Schedule a team sync by Friday")
    
    # Simulate conversation
    # print(agent.chat("Hi, my name is Sarah. I prefer meetings in the afternoon."))
    # print(agent.chat("Can we set up that team sync on Thursday morning?"))
    print("User: Hi, my name is Sarah. I prefer meetings in the afternoon.")
    print("Agent: Nice to meet you, Sarah. I will remember that you prefer afternoon meetings.")
    print("User: Can we set up that team sync on Thursday morning?")
    print("Agent: I know you prefer afternoon meetings, Sarah. Would you like me to schedule the team sync for Thursday afternoon instead to align with your preferences and our active goal?")\n
````
