## Responsible Agent


```python
class ResponsibleAgent:
    def __init__(self):
        self.blocked_keywords = ["confidential", "password", "ssn", "credit card"]
        self.disclaimer = "\n\n*Disclaimer: I am an AI assistant. Please verify critical information independently.*"

    def filter_input(self, user_input: str) -> bool:
        """Check for PII or sensitive data in user input."""
        for keyword in self.blocked_keywords:
            if keyword in user_input.lower():
                return False
        return True

    def generate_response(self, user_input: str) -> str:
        if not self.filter_input(user_input):
            return "I'm sorry, but for your privacy and security, I cannot process requests containing highly sensitive information (like passwords or SSNs)."
        
        # Simulate LLM response
        base_response = "I have scheduled the meeting for you."
        
        return base_response + self.disclaimer

if __name__ == "__main__":
    agent = ResponsibleAgent()
    print("Input: Schedule a meeting with John.")
    print(agent.generate_response("Schedule a meeting with John."))
    print("\nInput: My password is Password123, save it.")
    print(agent.generate_response("My password is Password123, save it."))\n
```
