```python
"""
Multi-Agent Workflow: Full Orchestrated Pipeline
Building on Task 2 setup, this implements the complete workflow graph.
"""
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.3)

class AgentNode:
    """Base class for all agent nodes in the workflow."""
    def __init__(self, name: str, system_prompt: str):
        self.name = name
        self.system_prompt = system_prompt

    def run(self, input_text: str) -> str:
        messages = [
            SystemMessage(content=self.system_prompt),
            HumanMessage(content=input_text)
        ]
        # response = llm(messages)
        # return response.content
        return f"[{self.name}] Processed: {input_text[:80]}..."


class MultiAgentOrchestrator:
    """Orchestrates a sequential pipeline of specialized agents."""

    def __init__(self):
        self.agents = {
            "researcher": AgentNode("ResearchAgent",
                "You are a research specialist. Gather and summarize relevant information on the given topic."),
            "analyst": AgentNode("AnalystAgent",
                "You are a data analyst. Extract key metrics and insights from the provided research summary."),
            "writer": AgentNode("WriterAgent",
                "You are a professional writer. Produce a polished, structured report from the analysis provided."),
            "qa": AgentNode("QAAgent",
                "You are a QA reviewer. Check the report for accuracy and professional quality. Approve or suggest edits.")
        }

    def run(self, user_request: str) -> dict:
        print(f"\n🚀 Starting Multi-Agent Workflow for: '{user_request}'\n")
        results = {}

        research = self.agents["researcher"].run(user_request)
        print(f"✅ Research complete")
        results["research"] = research

        analysis = self.agents["analyst"].run(research)
        print(f"✅ Analysis complete")
        results["analysis"] = analysis

        report = self.agents["writer"].run(analysis)
        print(f"✅ Report drafted")
        results["report"] = report

        final = self.agents["qa"].run(report)
        print(f"✅ QA approved")
        results["final"] = final

        return results


if __name__ == "__main__":
    orchestrator = MultiAgentOrchestrator()
    result = orchestrator.run("Analyze the global AI chip shortage and its impact on tech companies")
    print("\n=== Final Deliverable ===")
    for k, v in result.items():
        print(f"\n[{k.upper()}]\n{v}")
```
