## Delegation Logic


```python
"""
Task 4: LLM-Powered Inter-Agent Communication and Delegation
Demonstrates how the orchestrator delegates tasks using structured JSON messages.
"""
import json
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage, SystemMessage

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)

DELEGATION_PROMPT = """You are an AI orchestrator managing a team of specialized agents.
Given a user task, output a JSON delegation plan specifying which agents to call and in what order.
Available agents: researcher, analyst, writer, qa_reviewer

Output format:
{
  "task_breakdown": [
    {"step": 1, "agent": "researcher", "instruction": "..."},
    {"step": 2, "agent": "analyst", "instruction": "..."},
    ...
  ]
}"""

def generate_delegation_plan(user_task: str) -> dict:
    """Use LLM to generate a dynamic delegation plan."""
    messages = [
        SystemMessage(content=DELEGATION_PROMPT),
        HumanMessage(content=f"User Task: {user_task}")
    ]
    # response = llm(messages)
    # plan = json.loads(response.content)
    # Simulated plan
    plan = {
        "task_breakdown": [
            {"step": 1, "agent": "researcher", "instruction": "Research competitive landscape of cloud AI services"},
            {"step": 2, "agent": "analyst", "instruction": "Compare pricing and features of AWS, Azure, and GCP AI services"},
            {"step": 3, "agent": "writer", "instruction": "Write a 1-page comparison report for the CTO"},
            {"step": 4, "agent": "qa_reviewer", "instruction": "Verify all pricing figures and approve the report"}
        ]
    }
    return plan

def execute_delegation(plan: dict) -> list:
    """Execute each step in the delegation plan sequentially."""
    results = []
    for step in plan["task_breakdown"]:
        print(f"  Step {step['step']}: [{step['agent']}] → {step['instruction'][:60]}...")
        results.append({
            "step": step["step"],
            "agent": step["agent"],
            "output": f"[{step['agent']} completed task]"
        })
    return results

if __name__ == "__main__":
    user_task = "Compare cloud AI service providers for our enterprise procurement decision"
    print(f"User Task: {user_task}\n")
    print("Generating delegation plan...")
    plan = generate_delegation_plan(user_task)
    print(json.dumps(plan, indent=2))
    print("\nExecuting plan...")
    results = execute_delegation(plan)
    print(f"\n✅ All {len(results)} steps completed successfully")
```
