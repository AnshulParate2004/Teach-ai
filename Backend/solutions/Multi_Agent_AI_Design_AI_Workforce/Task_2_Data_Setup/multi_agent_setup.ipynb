# Task_2_Data_Setup
## Multi Agent Setup


```python
from crewai import Agent, Task, Crew, Process
from langchain_openai import ChatOpenAI

# ─── Initialize LLM ────────────────────────────────────────────────────────
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.3)

# ─── Define Agents ─────────────────────────────────────────────────────────
orchestrator = Agent(
    role="Project Orchestrator",
    goal="Break down user requests into sub-tasks and coordinate agents",
    backstory="You are a seasoned project manager who excels at delegating work to specialists.",
    verbose=True,
    llm=llm
)

researcher = Agent(
    role="Research Specialist",
    goal="Gather comprehensive information on the topic from reliable sources",
    backstory="You are an expert researcher with access to industry databases and web search.",
    verbose=True,
    llm=llm
)

analyst = Agent(
    role="Data Analyst",
    goal="Analyze gathered data and extract key insights and patterns",
    backstory="You are a data analyst who turns raw data into structured insights.",
    verbose=True,
    llm=llm
)

writer = Agent(
    role="Content Writer",
    goal="Produce clear, professional, and well-structured reports and summaries",
    backstory="You are a professional technical writer with expertise in business reports.",
    verbose=True,
    llm=llm
)

qa_agent = Agent(
    role="Quality Assurance Reviewer",
    goal="Review outputs for accuracy, consistency, and factual correctness",
    backstory="You are a meticulous reviewer who catches errors before delivery.",
    verbose=True,
    llm=llm
)

# ─── Define Tasks ───────────────────────────────────────────────────────────
research_task = Task(
    description="Research the current state of Electric Vehicle (EV) adoption in India. "
                "Focus on market size, key players, government policies, and growth trends.",
    agent=researcher,
    expected_output="A structured summary of EV adoption data with statistics and sources."
)

analysis_task = Task(
    description="Analyze the research findings and extract: market size, CAGR, "
                "top 3 players by market share, and 3 key growth drivers.",
    agent=analyst,
    expected_output="Structured analysis with key metrics and insights."
)

writing_task = Task(
    description="Write a professional 2-page market research report on EV adoption in India "
                "based on the research and analysis provided.",
    agent=writer,
    expected_output="A polished, well-structured 2-page report."
)

qa_task = Task(
    description="Review the draft report for factual accuracy, logical consistency, "
                "and professional quality. Provide a final approved version.",
    agent=qa_agent,
    expected_output="Final approved report with any corrections noted."
)

# ─── Assemble Crew ──────────────────────────────────────────────────────────
crew = Crew(
    agents=[orchestrator, researcher, analyst, writer, qa_agent],
    tasks=[research_task, analysis_task, writing_task, qa_task],
    process=Process.sequential,
    verbose=True
)

if __name__ == "__main__":
    print("=== Multi-Agent AI Workforce Demo ===")
    print("Running EV Market Research task...\n")
    # result = crew.kickoff()  # Uncomment with real API key
    # print(result)
    print("[Simulated Output]")
    print("Research Agent: Found 12 sources on EV adoption in India.")
    print("Analysis Agent: Market size $3.2B, CAGR 36%, Top players: Tata, Ola, Ather.")
    print("Writer Agent: Draft report completed (2 pages).")
    print("QA Agent: Report approved with minor corrections.")
    print("\nFinal Report delivered to user in 2.8 minutes.")
```
