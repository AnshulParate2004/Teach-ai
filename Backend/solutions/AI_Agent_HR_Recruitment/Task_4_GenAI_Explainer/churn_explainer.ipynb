## Churn Explainer


```python
from openai import OpenAI
client = OpenAI()

def explain_churn_risk(customer: dict) -> str:
    reasons = []
    if customer['usage_frequency'] < 2: reasons.append('very low product engagement')
    if customer['support_tickets'] > 5: reasons.append('high number of support issues')
    if customer['nps_score'] < 6: reasons.append('low satisfaction score')
    if customer['contract_type'] == 'Monthly': reasons.append('month-to-month contract')
    risk = 'HIGH' if len(reasons) >= 2 else 'MEDIUM' if reasons else 'LOW'
    reason_str = ', '.join(reasons) if reasons else 'no strong risk signals'
    return f'Churn Risk: {risk}. Key drivers: {reason_str}. Recommend proactive outreach.'

customers = [
    {'id': 'C001', 'usage_frequency': 1.2, 'support_tickets': 7, 'nps_score': 4, 'contract_type': 'Monthly'},
    {'id': 'C002', 'usage_frequency': 8.5, 'support_tickets': 1, 'nps_score': 9, 'contract_type': 'Annual'},
]
for c in customers:
    print(f"Customer {c['id']}: {explain_churn_risk(c)}")
```
