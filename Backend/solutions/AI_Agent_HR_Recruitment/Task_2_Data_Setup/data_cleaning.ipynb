# Task_2_Data_Setup
## Data Cleaning


```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 800
df = pd.DataFrame({
    'customer_id': range(1, n+1),
    'usage_frequency': np.random.uniform(0, 10, n),
    'support_tickets': np.random.randint(0, 15, n),
    'nps_score': np.random.randint(0, 11, n),
    'tenure_months': np.random.randint(1, 60, n),
    'contract_type': np.random.choice(['Monthly','Annual'], n, p=[0.4, 0.6]),
})
# Inject some missing values
df.loc[np.random.choice(df.index, 40), 'nps_score'] = np.nan
df.loc[np.random.choice(df.index, 20), 'usage_frequency'] = np.nan

# Define churn label
df['churned'] = ((df['usage_frequency'] < 2) & (df['nps_score'] < 6) | (df['support_tickets'] > 8)).astype(int)

print('Before cleaning:', df.isnull().sum().sum(), 'missing values')
df['nps_score'].fillna(df['nps_score'].median(), inplace=True)
df['usage_frequency'].fillna(df['usage_frequency'].mean(), inplace=True)
print('After cleaning:', df.isnull().sum().sum(), 'missing values')
df['contract_encoded'] = (df['contract_type'] == 'Annual').astype(int)
df.drop('contract_type', axis=1).to_csv('churn_clean.csv', index=False)
print('Cleaned dataset saved. Churn rate:', df['churned'].mean())
```
