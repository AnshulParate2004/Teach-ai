## Valuation Explainer


```python
from openai import OpenAI
client = OpenAI()

def explain_valuation(company, dcf_value, comps_value, audience='client'):
    if audience == 'client':
        return (f'Based on our analysis, {company} is worth approximately '
                f'${dcf_value:.0f}M to ${comps_value:.0f}M. '
                'This is based on its strong revenue growth (50% YoY) and '
                'comparing it to similar AI software companies in the market.')
    else:
        return (f'DCF implies ${dcf_value:.0f}M EV at 15% WACC. '
                f'Comps trading at 9x Revenue / 38x EBITDA implies ${comps_value:.0f}M. '
                'Recommend midpoint of $350M as fair value.')

print('Client Explanation:')
print(explain_valuation('TechNova Inc.', 312, 378, 'client'))
print('\nAnalyst Explanation:')
print(explain_valuation('TechNova Inc.', 312, 378, 'analyst'))
```
