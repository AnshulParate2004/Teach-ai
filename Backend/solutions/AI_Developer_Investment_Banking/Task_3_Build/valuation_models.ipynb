## Valuation Models


```python
import pandas as pd
import numpy as np

# DCF Valuation
revenue = 42  # $M
growth_rates = [0.40, 0.35, 0.30, 0.25, 0.20]
ebitda_margin = 0.22
discount_rate = 0.15
terminal_growth = 0.04

revenues = [revenue]
for g in growth_rates:
    revenues.append(revenues[-1] * (1 + g))

ebitdas = [r * ebitda_margin for r in revenues[1:]]
fcfs = [e * 0.75 for e in ebitdas]  # 75% conversion

pv_fcfs = [fcf / (1 + discount_rate)**(i+1) for i, fcf in enumerate(fcfs)]
terminal_value = fcfs[-1] * (1 + terminal_growth) / (discount_rate - terminal_growth)
pv_terminal = terminal_value / (1 + discount_rate)**5

enterprise_value_dcf = sum(pv_fcfs) + pv_terminal

# Comparable Company Valuation
ev_rev_multiple = 9.0  # peer median
ev_ebitda_multiple = 38.0
ev_comps_rev = revenue * ev_rev_multiple
ev_comps_ebitda = 8.4 * ev_ebitda_multiple

print('=== TechNova Valuation ===\n')
print(f'DCF Enterprise Value:      ${enterprise_value_dcf:.1f}M')
print(f'Comps (Revenue Multiple):  ${ev_comps_rev:.1f}M')
print(f'Comps (EBITDA Multiple):   ${ev_comps_ebitda:.1f}M')
print(f'Implied Valuation Range:   ${min(enterprise_value_dcf, ev_comps_rev):.0f}M - ${max(enterprise_value_dcf, ev_comps_ebitda):.0f}M')
```
