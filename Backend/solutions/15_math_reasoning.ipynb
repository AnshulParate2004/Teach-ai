# 15. Math Reasoning Agent with Structured JSON Output
**Industry:** Aerospace

Build an agent that solves step-by-step quantitative problems using a reasoning-focused LLM and outputs structured JSON.

```
!pip install langchain langchain-google-genai pydantic
```

```
from pydantic import BaseModel, Field
from langchain_google_genai import ChatGoogleGenerativeAI

class MathSolution(BaseModel):
    steps: list[str] = Field(description="Step by step reasoning")
    final_answer: float = Field(description="The final calculated numeric answer")

llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash")
structured_llm = llm.with_structured_output(MathSolution)

problem = "A plane burns 1200 gallons of fuel per hour. It flies for 3.5 hours. How much fuel did it burn?"
result = structured_llm.invoke(problem)

print(result.model_dump_json(indent=2))
```
