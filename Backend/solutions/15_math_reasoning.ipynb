```python
from pydantic import BaseModel, Field
import os
import dotenv
dotenv.load_dotenv(r"D:/Internship/Teach-ai/Backend/.env")
from langchain_openai import AzureChatOpenAI

class MathSolution(BaseModel):
    steps: list[str] = Field(description="Step by step reasoning")
    final_answer: float = Field(description="The final calculated numeric answer")

llm = AzureChatOpenAI(azure_endpoint=os.environ.get("AZURE_OPENAI_ENDPOINT"), api_key=os.environ.get("AZURE_OPENAI_API_KEY"), azure_deployment="gpt-4o", api_version=os.environ.get("AZURE_OPENAI_API_VERSION", "2024-12-01-preview"))
structured_llm = llm.with_structured_output(MathSolution)

problem = "A plane burns 1200 gallons of fuel per hour. It flies for 3.5 hours. How much fuel did it burn?"
result = structured_llm.invoke(problem)

print(result.model_dump_json(indent=2))
```
