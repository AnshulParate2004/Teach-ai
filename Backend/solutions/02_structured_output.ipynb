# 2. Structured Output with Pydantic
**Industry:** Retail & E-Commerce

Analyze a raw product description and return structured catalog data.

```
!pip install langchain langchain-google-genai pydantic python-dotenv
```

```
from pydantic import BaseModel, Field
from langchain_google_genai import ChatGoogleGenerativeAI
import os

# os.environ['GOOGLE_API_KEY'] = 'your-api-key-here'
llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash")

class ProductSchema(BaseModel):
    title: str = Field(description="The name of the product")
    category: str = Field(description="The category of the product (e.g. Electronics, Clothing)")
    price_tier: str = Field(description="'Budget', 'Mid-range', or 'Premium'")
    summary: str = Field(description="A short 1-sentence summary of the product")

structured_llm = llm.with_structured_output(ProductSchema)

description = "The new QuantumBook Pro 15 features an ultra-fast M4 chip, a stunning 4K OLED display, and costs $2499. Perfect for professional video editors."

result = structured_llm.invoke(description)
print(result.model_dump_json(indent=2))
```
