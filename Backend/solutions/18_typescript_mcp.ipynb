# 18. TypeScript MCP Server (Weather Data)
**Industry:** Tourism

Build a Model Context Protocol server in TypeScript/Node.js that exposes a 'get weather by city' tool.

> **Note:** This is a TypeScript project. The solution is provided as a reference `server.ts` file block here.

```python
"""
// server.ts
import { FastMCP } from 'fastmcp';
import { z } from 'zod';

const server = new FastMCP('Weather Server');

server.addTool({
  name: 'get_weather',
  description: 'Get weather for a city',
  parameters: z.object({
    city: z.string()
  }),
  execute: async (args) => {
    return `The weather in ${args.city} is sunny.`;
  }
});

server.start();
"""
print("Run this using: npm install fastmcp zod && npx ts-node server.ts")
```
