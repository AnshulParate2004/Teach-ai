```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate

llm = ChatOpenAI(model='gpt-3.5-turbo', temperature=0.2)

CUSTOM_PROMPT = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful customer service agent for TechCorp. '
               'Answer ONLY based on the provided context. '
               'Be concise, polite, and professional. '
               'If unsure, offer to connect the customer to a human agent.'),
    ('human', 'Context: {context}\n\nQuestion: {question}')
])

def custom_rag_answer(context: str, question: str) -> str:
    # chain = CUSTOM_PROMPT | llm
    # return chain.invoke({'context': context, 'question': question}).content
    return f'Based on our policy: {context[:100]}. Is there anything else I can help you with?'

ctx = 'Refund policy: 30-day money back guarantee with original receipt.'
q = 'Can I return a product after 2 weeks?'
print(f'Q: {q}\nA: {custom_rag_answer(ctx, q)}')
```
