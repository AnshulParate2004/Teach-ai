# 16. Dynamic Prompt Templating & Dual-LLM Comparison
**Industry:** Food & Beverages

Build a chatbot that sends the same user message to two different LLMs using dynamic prompt templates.

```
!pip install langchain langchain-google-genai
```

```
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
import os

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a food and beverage R&D assistant."),
    ("human", "{question}")
])

# Simulating two different models (e.g. Gemini Flash and Gemini Pro)
llm1 = ChatGoogleGenerativeAI(model="gemini-1.5-flash")
llm2 = ChatGoogleGenerativeAI(model="gemini-1.5-pro")

chain1 = prompt | llm1
chain2 = prompt | llm2

q = "What are the nutritional differences between oat milk and almond milk?"

res1 = chain1.invoke({"question": q})
res2 = chain2.invoke({"question": q})

print("--- Model 1 (Flash) ---\n", res1.content[:200], "...\n")
print("--- Model 2 (Pro) ---\n", res2.content[:200], "...")
```
