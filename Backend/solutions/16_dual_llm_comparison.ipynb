```python
from langchain_core.prompts import ChatPromptTemplate
import os
import dotenv
dotenv.load_dotenv(r"D:/Internship/Teach-ai/Backend/.env")
from langchain_openai import AzureChatOpenAI
import os

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a food and beverage R&D assistant."),
    ("human", "{question}")
])

# Simulating two different models (e.g. Gemini Flash and Gemini Pro)
llm1 = AzureChatOpenAI(azure_endpoint=os.environ.get("AZURE_OPENAI_ENDPOINT"), api_key=os.environ.get("AZURE_OPENAI_API_KEY"), azure_deployment="gpt-4o", api_version=os.environ.get("AZURE_OPENAI_API_VERSION", "2024-12-01-preview"))
llm2 = AzureChatOpenAI(azure_endpoint=os.environ.get("AZURE_OPENAI_ENDPOINT"), api_key=os.environ.get("AZURE_OPENAI_API_KEY"), azure_deployment="gpt-4o", api_version=os.environ.get("AZURE_OPENAI_API_VERSION", "2024-12-01-preview"))

chain1 = prompt | llm1
chain2 = prompt | llm2

q = "What are the nutritional differences between oat milk and almond milk?"

res1 = chain1.invoke({"question": q})
res2 = chain2.invoke({"question": q})

print("--- Model 1 (Flash) ---\n", res1.content[:200], "...\n")
print("--- Model 2 (Pro) ---\n", res2.content[:200], "...")
```
