# Task_2_Data_Setup
## Log Analysis


```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 2000
events = ['login_success','login_failure','file_access','port_scan','large_upload','file_encryption']
df = pd.DataFrame({
    'timestamp': pd.date_range('2024-01-01', periods=n, freq='5min'),
    'source_ip': [f'192.168.1.{i}' for i in np.random.randint(1,50,n)],
    'event_type': np.random.choice(events, n, p=[0.4,0.25,0.2,0.05,0.05,0.05]),
    'bytes_transferred': np.random.randint(0, 1000000, n),
    'user_id': [f'user_{i}' for i in np.random.randint(1,20,n)],
})
df.to_csv('network_logs.csv', index=False)
print('Created network_logs.csv:', df.shape)
print(df['event_type'].value_counts())
```
