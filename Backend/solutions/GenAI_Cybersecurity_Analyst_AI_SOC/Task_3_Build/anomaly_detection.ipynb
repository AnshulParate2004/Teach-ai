## Anomaly Detection


```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import LabelEncoder

events = ['login_success','login_failure','port_scan','large_upload','file_encryption']
n = 600
df = pd.DataFrame({
    'event_type': np.random.choice(events, n, p=[0.5,0.3,0.1,0.05,0.05]),
    'bytes': np.random.randint(0, 1000000, n),
    'port': np.random.choice([22,80,443,3389], n),
    'hour': np.random.randint(0, 24, n),
})
le = LabelEncoder()
df['event_enc'] = le.fit_transform(df['event_type'])
model = IsolationForest(contamination=0.05, random_state=42)
df['anomaly'] = model.fit_predict(df[['event_enc','bytes','port','hour']]) == -1
print(f'Anomalies detected: {df.anomaly.sum()} / {n}')
print(df[df.anomaly]['event_type'].value_counts())
```
