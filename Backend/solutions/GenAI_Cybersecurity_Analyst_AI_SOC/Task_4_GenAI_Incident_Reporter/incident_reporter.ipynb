## Incident Reporter


```python
from openai import OpenAI
client = OpenAI()

def generate_incident_report(event: dict) -> str:
    severity = 'HIGH' if event['type'] in ['port_scan','file_encryption'] else 'MEDIUM'
    return (f'INCIDENT REPORT\n'
            f'Summary: {event["type"]} from {event["ip"]} by {event["user"]}\n'
            f'Severity: {severity}\n'
            f'Action: Isolate IP. Review user activity. Escalate to Tier-2.')

event = {'ip': '192.168.1.47', 'type': 'port_scan', 'user': 'user_12', 'time': '03:42'}
print(generate_incident_report(event))
```
