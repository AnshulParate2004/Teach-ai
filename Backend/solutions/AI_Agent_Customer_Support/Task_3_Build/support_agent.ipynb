```python
from openai import OpenAI
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain.schema import Document
import json

client = OpenAI()

FAQS = [
    'How to reset password: Click Forgot Password on login page.',
    'Refund policy: 30-day window, submit via Account -> Billing.',
    'Add team member: Settings -> Team -> Invite Member.',
]

def detect_intent(query: str) -> str:
    q = query.lower()
    if any(w in q for w in ['password','login','access']): return 'account_access'
    if any(w in q for w in ['refund','charge','bill','invoice']): return 'billing'
    if any(w in q for w in ['add','invite','team','member']): return 'account_management'
    return 'general'

def answer_query(query: str) -> str:
    intent = detect_intent(query)
    responses = {
        'account_access': 'To reset your password, click Forgot Password on the login page and check your email.',
        'billing': 'For billing inquiries, go to Account -> Billing. Refunds take 3-5 business days.',
        'account_management': 'To add a team member: Settings -> Team -> Invite Member.',
        'general': 'I can help with password resets, billing, and account management. What do you need help with?',
    }
    return responses[intent]

queries = ['I forgot my password','Why was I charged extra?','How do I invite my colleague?']
for q in queries:
    print(f'Q: {q}\nA: {answer_query(q)}\n')
```
