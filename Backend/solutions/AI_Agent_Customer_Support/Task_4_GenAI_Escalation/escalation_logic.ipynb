# Task_4_GenAI_Escalation
## Escalation Logic


```python
from openai import OpenAI
client = OpenAI()

ESCALATION_TRIGGERS = ['angry','frustrated','lawsuit','terrible','useless','refund now','manager']

def needs_escalation(query: str, conversation_turns: int) -> bool:
    q = query.lower()
    if any(trigger in q for trigger in ESCALATION_TRIGGERS): return True
    if conversation_turns >= 3: return True
    return False

def generate_escalation_message() -> str:
    return ('I understand your frustration and I want to make sure you get the best help possible. '
            'I am connecting you to a senior support specialist right now. '
            'Your ticket ID is #TKT-4821. Expected wait time: 2 minutes.')

def handle_query(query: str, turn: int) -> dict:
    if needs_escalation(query, turn):
        return {'escalate': True, 'message': generate_escalation_message()}
    return {'escalate': False, 'message': f'AI response to: {query}'}

tests = [('This is terrible service!', 1), ('How do I reset password?', 1), ('Same issue again', 4)]
for query, turn in tests:
    result = handle_query(query, turn)
    print(f'Q ({turn}): {query}\nEscalate: {result["escalate"]} | {result["message"][:80]}\n')
```
