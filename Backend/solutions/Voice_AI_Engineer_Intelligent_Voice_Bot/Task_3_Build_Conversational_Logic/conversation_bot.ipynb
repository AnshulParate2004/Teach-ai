## Conversation Bot


```python
from openai import OpenAI
from langchain.memory import ConversationBufferMemory

client = OpenAI()

SYSTEM_PROMPT = """You are a helpful customer service voice bot for a telecom company called TeleCo.
You help users with billing queries, technical issues, and account management.
Keep responses short (1-2 sentences) since they will be spoken aloud.
If the issue requires human support, say you will transfer the call."""

class VoiceConversationBot:
    def __init__(self):
        self.memory = ConversationBufferMemory(return_messages=True)
        self.conversation_history = [{"role": "system", "content": SYSTEM_PROMPT}]

    def respond(self, user_text: str) -> str:
        """Process user input and return bot response text."""
        self.conversation_history.append({"role": "user", "content": user_text})

        response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=self.conversation_history,
            max_tokens=100,
            temperature=0.5
        )

        bot_reply = response.choices[0].message.content.strip()
        self.conversation_history.append({"role": "assistant", "content": bot_reply})
        return bot_reply

    def run_demo(self):
        """Run a simulated multi-turn conversation."""
        test_inputs = [
            "Hello, I have a problem with my bill.",
            "It shows ₹2,500 but I expected ₹1,200.",
            "Can you check my account please?"
        ]
        print("=== Voice Bot Conversation Demo ===\n")
        for user_input in test_inputs:
            print(f"User: {user_input}")
            # reply = self.respond(user_input)  # Uncomment with real API key
            simulated_replies = [
                "I'm sorry to hear that! Could you please provide your account number so I can look into the billing issue?",
                "I understand, that's a significant difference. Let me check if there are any additional charges or fees applied to your account.",
                "Of course! I can see your account now. It looks like a one-time activation fee was applied this month. Shall I explain the breakdown?"
            ]
            reply = simulated_replies[test_inputs.index(user_input)]
            print(f"Bot:  {reply}\n")


if __name__ == "__main__":
    bot = VoiceConversationBot()
    bot.run_demo()
```
