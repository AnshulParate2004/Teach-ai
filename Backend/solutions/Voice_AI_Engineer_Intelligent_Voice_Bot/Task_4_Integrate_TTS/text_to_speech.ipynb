## Text To Speech


```python
import os
import tempfile
from gtts import gTTS
from playsound import playsound

def text_to_speech(text: str, lang: str = "en", play: bool = True) -> str:
    """
    Convert text to speech audio using gTTS and optionally play it.

    Args:
        text: The text to convert to speech
        lang: Language code (default 'en' for English)
        play: Whether to play the audio immediately

    Returns:
        Path to the generated audio file
    """
    print(f"🔊 Generating speech for: '{text[:60]}...'")
    tts = gTTS(text=text, lang=lang, slow=False)

    # Save to temp file
    with tempfile.NamedTemporaryFile(suffix=".mp3", delete=False) as f:
        audio_path = f.name

    tts.save(audio_path)
    print(f"Audio saved to: {audio_path}")

    if play:
        playsound(audio_path)

    return audio_path


def text_to_speech_openai(text: str, voice: str = "alloy") -> str:
    """
    Higher quality TTS using OpenAI's TTS API.

    Args:
        text: Text to synthesize
        voice: Voice option - 'alloy', 'echo', 'fable', 'onyx', 'nova', 'shimmer'
    """
    from openai import OpenAI
    client = OpenAI()

    with tempfile.NamedTemporaryFile(suffix=".mp3", delete=False) as f:
        audio_path = f.name

    response = client.audio.speech.create(
        model="tts-1",
        voice=voice,
        input=text
    )
    response.stream_to_file(audio_path)
    return audio_path


if __name__ == "__main__":
    # Demo responses
    bot_responses = [
        "Hello! Welcome to TeleCo customer service. How can I help you today?",
        "I can see there is a one-time activation fee applied to your account this month.",
        "Is there anything else I can help you with?"
    ]

    print("=== TTS Demo ===")
    for response in bot_responses:
        print(f"\nText: {response}")
        # audio_path = text_to_speech(response, play=False)  # Uncomment to generate real audio
        print("  → Audio would be generated and played here (requires gTTS + playsound)")
```
