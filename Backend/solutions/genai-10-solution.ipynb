# Legal NDA Review Agent - Solution Notebook
**Industry:** Legal

This notebook contains a concrete implementation for the **Legal NDA Review Agent**. 
It demonstrates how to process specific business inputs and generate an AI response using the Groq API.

## Task Instructions
Your task is to build an agent that automatically reviews Non-Disclosure Agreements (NDAs). The user will input the NDA text, and the agent must compare it against the company playbook, flagging unacceptable clauses (like excessive durations or damages).

```
!pip install -q groq
import os
import json
from groq import Groq

# Initialize the Groq client (ensure your GROQ_API_KEY is set in your environment variables)
# os.environ["GROQ_API_KEY"] = "gsk_..." 
client = Groq()
```

## 1. Define Business Inputs
Here we define the mock data (e.g. resumes, policies, or client queries) that the agent will process.

```
nda_text = "The Receiving Party agrees not to disclose any Confidential Information for a period of 10 years. The Receiving Party shall pay $1,000,000 for any breach."
standard_playbook = "Confidentiality duration should not exceed 3 years. We do not accept liquidated damages clauses."
```

## 2. Prompt Design & System Instructions
Below is the core system prompt that dictates the agent's behavior and constraints.

```
system_prompt = """You are a Legal Review AI. Analyze the provided NDA text against our standard playbook. Flag any risky clauses and suggest revisions."""

user_prompt = f'''Playbook: {standard_playbook}\nNDA Text: {nda_text}'''

print("System Prompt:")
print(system_prompt)
print("\nUser Prompt:")
print(user_prompt)
```

## 3. Agent Execution
We will now call the Groq LLM API using the defined prompts to generate the response.

```
def run_agent(system_msg, user_msg):
    try:
        response = client.chat.completions.create(
            model="llama3-8b-8192",
            messages=[
                {"role": "system", "content": system_msg},
                {"role": "user", "content": user_msg}
            ],
            temperature=0.0
        )
        return response.choices[0].message.content
    except Exception as e:
        # Fallback mock response for testing without an API key
        return json.dumps({"status": "mock_response", "message": "API Key not provided. This is a fallback mock output."}, indent=2)

result = run_agent(system_prompt, user_prompt)
print("\n--- Agent Output ---")
print(result)
```
